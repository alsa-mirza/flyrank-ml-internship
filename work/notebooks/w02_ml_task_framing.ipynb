The lane that I have chosen is CTR Improvement by Search Position.
This problem is mainly based on prediction because the goal is to estimate the expected click-through rate (CTR) for web pages based on their SEO characteristics. The prediction helps in identifying pages that are performing below expectations and are needed to be improved.

The target variable in this case is CTR (Click Through Rate).
CTR represents the percentage of users who clicked on a search result after viewing it. This is the value that the machine learning model will learn to estimate.

A successful model should work effectively and should accurately estimate CTR for unseen pages. A lower prediction error means better performance. The predictions should help SEO specialists in identifying pages that deserve optimization.
